Atividade 1

```python
produto = input("Produto: ")
preco = float(input("Preço: "))
quantidade = int(input("Quantidade: "))
total = preco * quantidade


print("TIPOS DAS VARIÁVEIS")
print(f"produto ➔ {type(produto)}")
print(f"preco ➔ {type(preco)}")
print(f"quantidade ➔ {type(quantidade)}")
print(f"total ➔ {type(total)}")
```

Atividade 2

```python
n1 = 8
n2 = 7
n3 = 9
n4 = 6

media_errada = n1 + n2 + n3 + n4 / 4
media_correta = (n1 + n2 + n3 + n4) / 4

print(f"Média calculada incorretamente: {media_errada}")
print(f"Média calculada corretamente: {media_correta}")
```

Atividade 4

```python
temperatura = float(input("Digite a temperatura (°C): "))

if temperatura < 15:
    print("Frio")
elif temperatura < 25:
    print("Agradável")
else:
    print("Quente")
```

Atividade 5

```python
valor_compra = float(input("Digite o valor da compra (R$): "))

if valor_compra >= 200.00:
    valor_final = valor_compra * 0.90
else:
    valor_final = valor_compra

print(f"\nValor final a pagar: R$ {valor_final:.2f}")
```
