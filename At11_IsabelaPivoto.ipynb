Atividade 4

```python
soma = 0
contador = 0

for numero in range(2, 101, 2):
    soma = soma + numero
    contador = contador + 1
print(f"Quantidade de números pares encontrados: {contador}")
print(f"Soma total dos números pares de 2 a 100: {soma}")
```

Atividade 5

```python
numero = int(input('Digite um número: '))

for i in range(1, 11):
    resultado = numero * i

    print(f'{numero} x {i} = {resultado}')
```
