```python
i = 10

while i >= 1:
    print(i)

    i -= 1

print('Fim')
```

Atividade 3

```python
i = 1
soma = 0

while i <= 100:
    soma += i

    i += 1

print(f'Soma: {soma}')
```

Atividade 4

```python
idade = int(input('Digite sua idade: '))

while idade < 0 or idade > 120:
    print('Idade invalida. Digite um valor entre 0 e 120.')

    idade = int(input('Digite sua idade: '))

print(f'Idade aceita: {idade}')
```

Atividade 5

```python
numero = int(input('Digite um numero entre 1 e 10: '))

while numero < 1 or numero > 10:
    print('Numero invalido. Digite um valor entre 1 e 10.')
    numero = int(input('Digite um numero entre 1 e 10: '))
print(f'\nTabuada do {numero}:')

for i in range(1, 11):
    print(f'{numero} x {i} = {numero * i}')
```
