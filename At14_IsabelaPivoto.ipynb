Atividade 4

```python
produtos = [
    {"nome": "Caderno", "preco": 12.50, "qtd": 100},
    {"nome": "Caneta", "preco": 3.80, "qtd": 500},
    {"nome": "Mochila", "preco": 89.90, "qtd": 20}
]

valor_total_estoque = 0.0

for p in produtos:
    subtotal = p["preco"] * p["qtd"]
    valor_total_estoque += subtotal

print(f"Valor total do estoque: R$ {valor_total_estoque:.2f}")
```

Atividade 5

```python
comentarios = [
    "O atendimento foi excelente",
    "Gostei do preço",
    "ATENDIMENTO rápido e eficiente",
    "A entrega atrasou",
    "atendimento poderia melhorar"
]

contador = 0

for comentario in comentarios:
    if "atendimento" in comentario.lower():
        contador += 1

print(f"Total de comentários que mencionam 'atendimento': {contador}")
```
