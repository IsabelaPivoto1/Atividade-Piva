Atividade 3

```python
produtos = [
    {"nome": "Caderno", "preco": 12.50, "qtd": 100},
    {"nome": "Caneta", "preco": 3.80, "qtd": 500},
    {"nome": "Mochila", "preco": 89.90, "qtd": 20}
]
```

Atividade 4

```python
total_estoque = 0

for produto in produtos:
    subtotal = produto["preco"] * produto["qtd"]
    total_estoque += subtotal

print(f"Valor total do estoque: R$ {total_estoque:.2f}")
```

Atividade 5

```python
comentarios = [
    "O atendimento foi excelente",
    "Gostei do preço",
    "ATENDIMENTO rápido e eficiente",
    "A entrega atrasou",
    "atendimento poderia melhorar"
]

contador = 0

for comentario in comentarios:
    if "atendimento" in comentario.lower():
        contador += 1

print(f"Total de menções a 'atendimento': {contador}")
```
