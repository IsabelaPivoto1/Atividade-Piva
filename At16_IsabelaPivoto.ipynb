```python
def dobrar(n):
    return n * 2

assert dobrar(2) == 4, 'Esperava 4'
assert dobrar(0) == 0, 'Esperava 0'
assert dobrar(-3) == -6, 'Esperava -6'

print(dobrar(5))
print('Testes passaram!')
```

Atividade 4

```python
def calcular_media_simples(n1, n2, n3):
    return (n1 + n2 + n3) / 3

assert calcular_media_simples(10, 8, 6) == 8, 'Média de 10, 8, 6 deveria ser 8'
assert calcular_media_simples(7, 7, 7) == 7, 'Média de notas iguais deveria ser a nota'
assert calcular_media_simples(0, 0, 0) == 0, 'Média de zeros é zero'

print(calcular_media_simples(8.0, 7.5, 9.0))
print('Testes passaram!')
```

Atividade 5


```python
def classificar_aluno(media):
    if media >= 7:
        return 'Aprovado'
    elif media >= 5:
        return 'Recuperação'
    else:
        return 'Reprovado'

assert classificar_aluno(8.5) == 'Aprovado', 'Esperava Aprovado'
assert classificar_aluno(6.2) == 'Recuperação', 'Esperava Recuperação'
assert classificar_aluno(3.0) == 'Reprovado', 'Esperava Reprovado'

assert classificar_aluno(7.0) == 'Aprovado', 'Esperava Aprovado para 7.0'
assert classificar_aluno(5.0) == 'Recuperação', 'Esperava Recuperação para 5.0'
assert classificar_aluno(4.9) == 'Reprovado', 'Esperava Reprovado para 4.9'
print('Todos os testes passaram com sucesso!')
```

```python
def calcular_ticket_medio(faturamento, pedidos):
    if pedidos == 0:
        return 0.0
    return faturamento / pedidos

assert calcular_ticket_medio(1200, 12) == 100.0, 'Esperava 100.0'[cite: 6]
assert calcular_ticket_medio(0, 10) == 0.0, 'Esperava 0.0 para faturamento zero'[cite: 6]
assert calcular_ticket_medio(1000, 0) == 0.0, 'Esperava 0.0 para pedidos zero'[cite: 6]

print(calcular_ticket_medio(1200, 12))
print('Testes passaram!')
```
