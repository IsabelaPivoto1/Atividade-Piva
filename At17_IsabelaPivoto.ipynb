Atividade 2

```python
import math

numero = float(input("Digite um número decimal: "))

raiz = math.sqrt(numero)
arredondado_baixo = math.floor(numero)
arredondado_cima = math.ceil(numero)

print(f"Raiz quadrada (duas casas decimais): {raiz:.2f}")
print(f"Valor arredondado para baixo: {arredondado_baixo}")
print(f"Valor arredondado para cima: {arredondado_cima}")
```

Atividade 4

```python
import random
import statistics

random.seed(42)

notas = []
for _ in range(20):
    nota = random.uniform(0, 10)
    notas.append(nota)

media = statistics.mean(notas)
mediana = statistics.median(notas)
desvio_padrao = statistics.pstdev(notas)

print("RELATÓRIO DE NOTAS ")
print("Notas geradas:")
for n in notas:
    print(f"{n:.2f}", end=" ")
print("\n")

print(f"Média: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Desvio Padrão Populacional: {desvio_padrao:.2f}")
```

Atividade 5

```python
import random
import statistics
import math

lancamentos = [random.randint(1, 6) for _ in range(1000)]

media = statistics.mean(lancamentos)
mediana = statistics.median(lancamentos)
desvio_padrao = statistics.pstdev(lancamentos)

media_proxima = math.isclose(media, 3.5, abs_tol=0.2)

print("--- RELATÓRIO DOS LANÇAMENTOS ---")
print(f"Média: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Desvio Padrão Populacional: {desvio_padrao:.2f}")
print(f"A média está próxima de 3.5? {'Sim' if media_proxima else 'Não'}\n")

print("--- HISTOGRAMA TEXTUAL ---")
for face in range(1, 7):
    contagem = lancamentos.count(face)
    barra = "█" * (contagem // 10)
    print(f"Face {face}: {barra} ({contagem} vezes)")
```

Atividade 6

```python
from datetime import datetime, date

data_texto = input("Digite a data de vencimento (DD/MM/AAAA): ")
vencimento = datetime.strptime(data_texto, "%d/%m/%Y").date()

hoje = date.today()

diferenca = vencimento - hoje
dias = diferenca.days

if dias > 0:
    print(f"Faltam {dias} dias para o vencimento.")
elif dias < 0:
    print(f"O vencimento já passou há {abs(dias)} dias.")
else:
    print("O vencimento é hoje!")
```
