Atividade 1

```python
def quadrado(n):
    return n * n

resultado = quadrado(5)
dobro_do_quadrado = resultado * 2

print(resultado)
print(dobro_do_quadrado)

assert quadrado(5) == 25
assert quadrado(0) == 0
assert quadrado(-3) == 9
```

```python
def calcular_desconto(preco, percentual):
    return preco * percentual

valor = calcular_desconto(100, 0.10)
assert calcular_desconto(100, 0.10) == 10
assert calcular_desconto(200, 0.25) == 50
assert calcular_desconto(0, 0.10) == 0
```

Atividade 3

```python
def classificar(media):
    if media >= 7:
        return "Aprovado"
    elif media >= 5:
        return "Exame"
    return "Reprovado"

assert classificar(7.0) == "Aprovado"
assert classificar(6.9) == "Exame"
assert classificar(5.0) == "Exame"
assert classificar(4.9) == "Reprovado"
```

Atividade 4

```python
def calcular_comissao(total, taxa=0.05):
    return total * taxa

assert calcular_comissao(1000) == 50
assert calcular_comissao(0) == 0
assert calcular_comissao(2000, 0.10) == 200
assert calcular_comissao(1500, 0.04) == 60
```

Atividade 5

```python
import math

def calcular_media(n1, n2, n3):
    return (n1 + n2 + n3) / 3

def classificar_situacao(media):
    if media >= 7:
        return "Aprovado"
    elif media >= 5:
        return "Exame"
    return "Reprovado"

assert math.isclose(calcular_media(7, 8, 9), 8)
assert calcular_media(0, 0, 0) == 0
assert calcular_media(10, 10, 10) == 10

assert classificar_situacao(7.0) == "Aprovado"
assert classificar_situacao(5.0) == "Exame"
assert classificar_situacao(4.9) == "Reprovado"
```

Atividade 6

```python
import statistics as stats

valores = [1200, 1500, 1700, 1600, 1800]

media = stats.mean(valores)
desvio_populacional = stats.pstdev(valores)

print(media)
print(f'{desvio_populacional:.2f}')

assert media == 1560
assert round(desvio_populacional, 2) == 205.91
```

Atividade 7

```python
def melhor_aluno(turma):
    """Retorna o nome e a média do aluno com maior média."""
    if len(turma) == 0:
        return None, None

    melhor_nome = turma[0]['nome']
    melhor_media = turma[0]['media']

    for aluno in turma:
        if aluno['media'] > melhor_media:
            melhor_nome = aluno['nome']
            melhor_media = aluno['media']

    return melhor_nome, melhor_media

turma = [
    {"nome": "Ana", "media": 8.0},
    {"nome": "Bruno", "media": 6.5},
    {"nome": "Carla", "media": 9.0}
]

nome, media = melhor_aluno(turma)

assert nome == "Carla"
assert media == 9.0
assert melhor_aluno([]) == (None, None)
```
