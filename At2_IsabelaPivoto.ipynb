Exercício 7

```python
# --- Dados do Problema ---
taxa_kbps = 128            # 128 kilobits por segundo
duracao_minutos = 3        # 3 minutos

# --- 1. Conversões de Tempo e Bits ---
duracao_segundos = duracao_minutos * 60
total_bits = taxa_kbps * 1000 * duracao_segundos

# --- 2. Conversão para Bytes ---
total_bytes = total_bits // 8

# --- 3. Conversão para Megabytes ---
tamanho_mb_decimal = total_bytes / 1_000_000     # Padrão comercial (1 MB = 1.000.000 Bytes)
tamanho_mib_binario = total_bytes / (1024 * 1024)  # Padrão de sistemas (1 MiB = 1.048.576 Bytes)

# --- Exibição dos Resultados ---
print(f"--- RESULTADO DO CÁLCULO ---")
print(f"Duração total: {duracao_segundos} segundos")
print(f"Total em Bits: {total_bits:,} bits")
print(f"Tamanho em Bytes: {total_bytes:,} bytes")
print(f"Tamanho em Megabytes (Decimal): {tamanho_mb_decimal:.2f} MB")
```

Na minha visão, o sistema não tenta criar uma cor única do zero para cada ponto da foto. Em vez disso, ele usa o modelo RGB (Red, Green, Blue - Vermelho, Verde e Azul). Cada pixel individual da foto é formado pela mistura dessas três cores de luz

Para definir o tom exato de cada uma dessas três cores, o sistema atribui um valor numérico para a intensidade de luz de cada canal. Se usarmos a escala padrão de 8 bits por cor, a variação da intensidade vai de 0 (totalmente escuro/desligado) a 255 (intensidade máxima da cor).

A Conversão para a Sequência de Bits, é aqui que os números viram dados binários puros na memória do computador. Como cada uma das 3 cores precisa de 1 byte (8 bits) para armazenar um número de 0 a 255, cada pixel individual da fotografia consome exatamente 24 bits (3 bytes)

Exercícios 14

Bit é como uma única letra: É a menor unidade possível. Sozinha, uma letra solta (como um "A" ou um "B") não consegue transmitir uma ideia completa ou um significado claro. No mundo digital, o bit é essa menor unidade — uma lâmpada que só pode estar acesa (1) ou apagada (0).

Byte é como uma palavra inteira: É quando você junta 8 letras (bits) para formar algo que realmente faz sentido e representa um significado completo. No computador, é preciso juntar 8 bits para formar 1 byte — e 1 byte é o espaço necessário para guardar uma letra, um número ou um espaço em branco (como a letra "A" ou o código do espaço que vimos antes).

A IA traduziu os conceitos abstratos de como funciona bit e byte e suas diferenças, para que seja um jeito mais didático e entendivel para aprender sobre ambos.
