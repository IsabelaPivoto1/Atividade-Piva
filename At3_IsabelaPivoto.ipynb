Exercício 15

```python
def calcular_tamanho_imagem_rgb(largura: int, altura: int):
    # Modelo RGB: 3 bytes por pixel (1 byte para Red, 1 para Green, 1 para Blue)
    BYTES_POR_PIXEL = 3

    # Convenção binária (1 KB = 1.024 bytes / 1 MB = 1.024 KB = 1.048.576 bytes)
    BYTES_POR_KB = 1024
    BYTES_POR_MB = 1024 * 1024  # 1.048.576 bytes

    # 1. Quantidade total de pixels da imagem
    total_pixels = largura * altura

    # 2. Tamanho total em bytes (multiplica pixels por 3 bytes/pixel)
    tamanho_bytes = total_pixels * BYTES_POR_PIXEL

    # 3. Conversão para KB e MB na base 1024
    tamanho_kb = tamanho_bytes / BYTES_POR_KB
    tamanho_mb = tamanho_bytes / BYTES_POR_MB

    return total_pixels, tamanho_bytes, tamanho_kb, tamanho_mb


# --- Exemplo de Teste (Imagem Full HD: 1920x1080) ---
largura_img = 1920
altura_img = 1080

pixels, b_total, kb_total, mb_total = calcular_tamanho_imagem_rgb(
    largura_img, altura_img
)

print(f"Resolução: {largura_img} x {altura_img}")
print(f"Total de Pixels: {pixels:,} pixels")
print(f"Tamanho em Bytes: {b_total:,} B")
print(f"Tamanho em KB (1024 B): {kb_total:,.2f} KB")
print(f"Tamanho em MB (1024 KB): {mb_total:,.2f} MB")
```
