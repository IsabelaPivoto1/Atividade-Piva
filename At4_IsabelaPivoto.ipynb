Exercício 5

```python
# 1. ENTRADA DE DADOS
# Lista simples com as notas de cada atendente
notas_carlos = [5, 4]
notas_ana = [3, 2]
notas_bruno = [4, 5]

# 2. PROCESSAMENTO
# Calcular a média de cada um (soma das notas dividida pela quantidade)
media_carlos = sum(notas_carlos) / len(notas_carlos)
media_ana = sum(notas_ana) / len(notas_ana)
media_bruno = sum(notas_bruno) / len(notas_bruno)

# 3. SAÍDA
print("--- MÉDIAS DOS ATENDENTES ---")
print("Carlos:", media_carlos)
print("Ana:", media_ana)
print("Bruno:", media_bruno)

print("\n--- ATENDENTES ABAIXO DA META (< 4.0) ---")

# Verifica um por um quem ficou abaixo de 4.0
if media_carlos < 4.0:
    print("- Carlos")

if media_ana < 4.0:
    print("- Ana")

if media_bruno < 4.0:
    print("- Bruno")
```

Exercicio 9

```python

def verificar_resultado(nota1, nota2, nota3):

    # 1. Processamento: Calcular a média
    media = (nota1 + nota2 + nota3) / 3

    # 2. Processamento: Verificar as condições dadas no exercício
    if media >= 5.0:
        status = "Aprovado"
    elif media >= 3.0:  # Como já verificou se é >= 5, aqui a média é necessariamente menor que 5
        status = "Recuperação"
    else:
        status = "Reprovado"

    # 3. Saída: Mostrar o resultado formatado com 1 casa decimal
    print(f"Média: {media:.1f} -> Saída esperada: {status}")

print("--- CASO A (Notas: 10, 8 e 9) ---")
verificar_resultado(10, 8, 9)

print("\n--- CASO B (Notas: 4, 4 e 4) ---")
verificar_resultado(4, 4, 4)

print("\n--- CASO C (Notas: 2, 1 e 3) ---")
verificar_resultado(2, 1, 3)
```
