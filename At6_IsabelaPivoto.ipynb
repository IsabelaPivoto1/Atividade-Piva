Atividade 1

```python

print(" A) ESTRUTURA DE SEQUÊNCIA ")
nome = input("Digite o seu nome: ")
print("Olá,", nome, "! Seja bem-vindo(a)!")
print()


print(" B) ESTRUTURA DE DECISÃO ")
media = float(input("Digite a nota média do aluno: "))

if media >= 5:
    print("Resultado: Aluno Aprovado!")
else:
    print("Resultado: Aluno Reprovado.")
print()


print(" C) ESTRUTURA DE REPETIÇÃO ")
print("Exibindo os números de 1 a 10:")
for numero in range(1, 11):
    print(numero)
print()


print(" D) ESTRUTURA DE REPETIÇÃO COM CONDIÇÃO -- WHILE ")
print("Digite a senha de acesso:")
senha_correta = "1234"
senha_digitada = input("Digite a senha de acesso: ")

while senha_digitada != senha_correta:
    print("Senha incorreta! Tente novamente.")
    senha_digitada = input("Digite a senha: ")

print("Acesso permitido! Senha correta.")
```

Atividade 10

```python
import math

opcao = -1

while opcao != 0:
    print("\n MENU INTERATIVO ")
    print("1. Calcular área do círculo")
    print("2. Calcular área do retângulo")
    print("0. Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        raio = float(input("Digite o raio do círculo: "))
        area_circulo = math.pi * (raio ** 2)
        print(f"➔ A área do círculo é: {area_circulo:.2f}")

    elif opcao == 2:
        base = float(input("Digite a base do retângulo: "))
        altura = float(input("Digite a altura do retângulo: "))
        area_retangulo = base * altura
        print(f"➔ A área do retângulo é: {area_retangulo:.2f}")

    elif opcao == 0:
        print("Saindo do programa. Até logo!")

    else:
        print("Opção inválida! Escolha 1, 2 ou 0.")
```
