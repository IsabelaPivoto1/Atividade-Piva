Atividade 5

```python
faturamento_total = float(input("Digite o faturamento total (R$): "))
quantidade_pedidos = int(input("Digite a quantidade de pedidos: "))

ticket_medio = faturamento_total / quantidade_pedidos
print(f"Ticket médio: R$ {ticket_medio:.2f}")
```

Atividade 6

```python
celsius = float(input("Digite a temperatura em graus Celsius (°C): "))

fahrenheit = celsius * 9 / 5 + 32
print(f"{celsius:.1f}°C equivalem a {fahrenheit:.1f}°F")
```
