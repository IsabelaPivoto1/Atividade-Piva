Atividade 1

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

soma = num1 + num2
subtracao = num1 - num2
multiplicacao = num1 * num2

if num2 != 0:
    divisao = num1 / num2
else:
    divisao = "Impossível dividir por zero"

print("\n RESULTADOS DAS OPERAÇÕES ")
print(f"Soma: {soma}")
print(f"Subtração: {subtracao}")
print(f"Multiplicação: {multiplicacao}")
print(f"Divisão: {divisao}")
```

Atividade 2

```python
total_produtos = int(input("Digite a quantidade total de produtos: "))
produtos_por_caixa = int(input("Digite quantos produtos cabem em cada caixa: "))


caixas_completas = total_produtos // produtos_por_caixa
produtos_restantes = total_produtos % produtos_por_caixa

print(f"\n➔ Caixas completas montadas: {caixas_completas}")
print(f"➔ Produtos restantes (sem formar caixa): {produtos_restantes}")
```

Atividade 4

```python
preco_original = float(input("Digite o preço original do produto (R$): "))
percentual = float(input("Digite o percentual de desconto (0 a 50%): "))


if preco_original <= 0:
    print("Erro: O preço original deve ser maior que zero!")

elif percentual < 0 or percentual > 50:
    print("Erro: O percentual de desconto deve estar entre 0% e 50%[cite: 25]!")

else:
    valor_desconto = preco_original * (percentual / 100)
    preco_final = preco_original - valor_desconto

    print("RESULTADO DO CÁLCULO")
    print(f"Preço original: R$ {preco_original:.2f}")
    print(f"Desconto aplicado: R$ {valor_desconto:.2f} ({percentual}%)")
    print(f"Preço final a pagar: R$ {preco_final:.2f}")
```
