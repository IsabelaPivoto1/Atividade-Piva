Atividade 4

```python
valor_compra = float(input("Digite o valor da compra (R$): "))

if valor_compra > 200.00:
    valor_final = valor_compra * 0.90
else:
    valor_final = valor_compra

print("RESUMO DA COMPRA ")
print("-" * 20)
print(f"Valor original: R$ {valor_compra:.2f}")
print(f"Valor final a pagar: R$ {valor_final:.2f}")
```

Atividade 5

```python
total_compras = float(input("Digite o total comprado no mês (R$): "))

if total_compras >= 1000.00:
    print("Categoria: Cliente Ouro")
elif total_compras >= 500.00:
    print("Categoria: Cliente Prata")
else:
    print("Categoria: Cliente Bronze")
```

Atividade 6

```python
meta = float(input("Digite a meta mensal de vendas (R$): "))
vendido = float(input("Digite o valor total vendido (R$): "))

diferenca = vendido - meta

diferenca_positiva = abs(diferenca)

if vendido >= meta:
    print("\nParabéns! Meta atingida ou superada!")
    print(f"Você superou a meta em R$ {diferenca_positiva:.2f}")
else:
    print("\nMeta não atingida.")
    print(f"Faltaram R$ {diferenca_positiva:.2f} para alcançar a meta.")
```
